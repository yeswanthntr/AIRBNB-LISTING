##. Price Distribution

Let's examine how Airbnb listing prices are distributed and identify whether most listings fall within a particular price range.





```python
plt.figure(figsize=(10,6))
sns.histplot(df['price'],bins=20,color='red')
plt.xlabel('price')
plt.ylabel('frequency')
plt.title('price distribution')
plt.show()
```



### Observation

The price distribution is highly right-skewed. Most Airbnb listings are concentrated in the lower price range, while a small number of listings have very high prices, creating a long right tail. This indicates the presence of high-price outliers in the dataset.



ROOMS BOOKING

Which type of Airbnb room has the highest number of listings?


```python
plt.figure(figsize=(10,6))
sns.countplot(data=df,x='room_type')
plt.xlabel('room types')
plt.ylabel('frequency')
plt.title('room types')
plt.show()
```

### OBSERVATION


THE CHARTS SHOWS THAT THERE ARE TWO TYPES OF ROOMS ARE MOST FREQUENTLY BOOKED SHARED ROOMS BOOKING ARE VERY LOW COMPARLY TO OTHERS


Which room type has the highest average number of reviews?

```python
plt.figure(figsize=(8,5))

rating_data = df.groupby('room_type')['number_of_reviews'].mean().reset_index()

sns.lineplot(
    data=rating_data,
    x='room_type',
    y='number_of_reviews',
    marker='o'
)

plt.title('Average Number of Reviews by Room Type')
plt.xlabel('Room Type')
plt.ylabel('Average Number of Reviews')
plt.show()
```

### Observation

The line chart compares the average number of reviews received by different room types. The room type with the highest average number of reviews shows greater review activity, while the room type with the lowest average has comparatively less review activity.

 Top 20 Neighbourhoods by Number of Listings

```python
top_10_neighbourhoods = (
    df['neighbourhood']
    .value_counts()
    .head(20)
    .reset_index()
)

top_10_neighbourhoods.columns = ['neighbourhood', 'listing_count']

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_10_neighbourhoods,
    x='listing_count',
    y='neighbourhood'
)

plt.title('Top 10 Neighbourhoods by Number of Airbnb Listings')
plt.xlabel('Number of Listings')
plt.ylabel('Neighbourhood')
plt.show()
```

### Observation

The chart shows the top 20 neighbourhoods with the highest number of Airbnb listings. A small number of neighbourhoods account for a large share of the listings, indicating that Airbnb activity is concentrated in specific areas.

```python
numeric_cols = df.select_dtypes(include='number').columns

plt.figure(figsize=(15, 12))

for i, col in enumerate(numeric_cols, 1):
    plt.subplot(3, 4, i) # Changed from (3, 3, i) to (3, 4, i)
    sns.boxplot(y=df[col])
    plt.title(f'Boxplot - {col}')
    plt.ylabel(col)

plt.tight_layout()
plt.show()
```

## Outlier Detection Using Boxplots

Boxplots were created using subplots to identify potential outliers across all numerical variables. Values outside the whiskers represent potential outliers in the respective variables.
